```python
from pathlib import Path
import sqlparse
spark.conf.set("spark.sql.debug.maxToStringFields", "2000")

# Local notebook; for a hosted notebook, upload the SQL and change this path.
path = Path(
    r"osdu_well_wellbore_iceberg.sql"
)

script = path.read_text(encoding="utf-8")
statements = sqlparse.split(sqlparse.format(script, strip_comments=True))

for index, statement in enumerate(statements, start=1):
    if statement.strip():
        print(f"Executing {index}/{len(statements)}")
        spark.sql(statement)

print("Finished creating tables.")

spark.sql("SHOW TABLES IN osdu_iceberg.master_data").show(truncate=False)

spark.sql("""
    DESCRIBE TABLE osdu_iceberg.master_data.Wellbore
""").show(200, truncate=False)
```
